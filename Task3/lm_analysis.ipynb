```python
# 📘 TEXT INSIGHTS ANALYSIS BLOCK

# Ensure required libraries are installed
!pip install textblob textstat --quiet

# Import necessary libraries
from textblob import TextBlob
from textstat import flesch_reading_ease
from collections import Counter
from nltk.tokenize import word_tokenize, sent_tokenize

# --- 1️⃣ Analyze readability ---
readability_score = flesch_reading_ease(generated_text)
print("📗 Readability (Flesch Score):", readability_score)
if readability_score > 80:
    print("➡️ Very easy to read (conversational language)")
elif readability_score > 60:
    print("➡️ Standard readability (good balance)")
elif readability_score > 30:
    print("➡️ Fairly difficult to read (complex sentences)")
else:
    print("➡️ Difficult to read (academic or technical language)")

print("\n--------------------------------------------------")

# --- 2️⃣ Sentiment Analysis ---
sentiment = TextBlob(generated_text).sentiment
print("💬 Sentiment Analysis:")
print(f"   Polarity: {sentiment.polarity:.2f} (−1 = negative, +1 = positive)")
print(f"   Subjectivity: {sentiment.subjectivity:.2f} (0 = objective, 1 = opinionated)")

print("\n--------------------------------------------------")

# --- 3️⃣ Frequent Words ---
words = word_tokenize(generated_text.lower())
freq = Counter([w for w in words if w.isalpha()])
print("🔠 Top 10 Frequent Words:")
for word, count in freq.most_common(10):
    print(f"   {word}: {count}")

print("\n--------------------------------------------------")

# --- 4️⃣ Sentence & Word Stats ---
sentences = sent_tokenize(generated_text)
print("📈 Text Statistics:")
print("   Total Sentences:", len(sentences))
print("   Total Words:", len(words))
print("   Average Words per Sentence:", round(len(words)/len(sentences), 2))
```

```python
# 📊 GPT-2 Text Insights + Visualization Dashboard

# Install required libraries (run once)
!pip install textblob textstat matplotlib seaborn wordcloud --quiet

# Import libraries
from textblob import TextBlob
from textstat import flesch_reading_ease
from collections import Counter
from nltk.tokenize import word_tokenize, sent_tokenize
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud

# --- 1️⃣ Readability ---
readability_score = flesch_reading_ease(generated_text)
print("📗 Readability (Flesch Score):", readability_score)
if readability_score > 80:
    print("➡️ Very easy to read (conversational language)")
elif readability_score > 60:
    print("➡️ Standard readability (good balance)")
elif readability_score > 30:
    print("➡️ Fairly difficult to read (complex sentences)")
else:
    print("➡️ Difficult to read (academic or technical language)")

print("\n--------------------------------------------------")

# --- 2️⃣ Sentiment Analysis ---
sentiment = TextBlob(generated_text).sentiment
print("💬 Sentiment Analysis:")
print(f"   Polarity: {sentiment.polarity:.2f} (−1 = negative, +1 = positive)")
print(f"   Subjectivity: {sentiment.subjectivity:.2f} (0 = objective, 1 = opinionated)")

print("\n--------------------------------------------------")

# --- 3️⃣ Frequent Words ---
words = word_tokenize(generated_text.lower())
freq = Counter([w for w in words if w.isalpha()])
top_words = freq.most_common(10)

print("🔠 Top 10 Frequent Words:")
for word, count in top_words:
    print(f"   {word}: {count}")

print("\n--------------------------------------------------")

# --- 4️⃣ Sentence & Word Stats ---
sentences = sent_tokenize(generated_text)
print("📈 Text Statistics:")
print("   Total Sentences:", len(sentences))
print("   Total Words:", len(words))
print("   Average Words per Sentence:", round(len(words)/len(sentences), 2))

print("\n--------------------------------------------------")

# --- 5️⃣ Word Frequency Bar Chart ---
plt.figure(figsize=(8,4))
sns.barplot(x=[w for w, _ in top_words], y=[c for _, c in top_words], palette="Blues_d")
plt.title("Top 10 Most Frequent Words", fontsize=14)
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.show()

# --- 6️⃣ Word Cloud ---
wordcloud = WordCloud(width=800, height=400, background_color='white').generate(generated_text)
plt.figure(figsize=(10,5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis("off")
plt.title("Word Cloud of Generated Text", fontsize=14)
plt.show()

# --- 7️⃣ Sentiment Visualization (Optional Gauge) ---
plt.figure(figsize=(5, 1))
plt.barh(["Sentiment Polarity"], [sentiment.polarity], color="lightgreen" if sentiment.polarity > 0 else "salmon")
plt.xlim(-1, 1)
plt.title("Sentiment Polarity Meter")
plt.show()
```

```python
# 🤖 HUMAN vs AI TEXT SIMILARITY CHECK

# Install Sentence Transformers (run once)
!pip install sentence-transformers --quiet

from sentence_transformers import SentenceTransformer, util

# Load model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Define your GPT-2 output and a sample human-written paragraph for comparison
human_text = """
Artificial intelligence is transforming industries by automating tasks,
enhancing decision-making, and creating new opportunities. As algorithms
improve, machines are learning to understand human language, generate creative
content, and assist professionals in various fields.
"""

# Encode both texts
embeddings1 = model.encode(generated_text, convert_to_tensor=True)
embeddings2 = model.encode(human_text, convert_to_tensor=True)

# Compute cosine similarity
similarity = util.cos_sim(embeddings1, embeddings2).item()

print("🤝 AI vs Human Text Similarity Score:", round(similarity, 3))
if similarity > 0.75:
    print("✅ Very close to human writing style")
elif similarity > 0.5:
    print("⚖️ Moderately similar to human text")
else:
    print("⚠️ Quite different from human writing style")

# Optional visualization (bar)
import matplotlib.pyplot as plt
plt.figure(figsize=(5,1))
plt.barh(["Similarity"], [similarity], color="lightblue")
plt.xlim(0,1)
plt.title("AI vs Human Similarity Score")
plt.show()
```
